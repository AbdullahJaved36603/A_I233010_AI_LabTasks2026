## Task 3

```python
titanic=sns.load_dataset("titanic")
titanic.isnull().sum()
print(titanic.head())
titanic["age"] = titanic["age"].fillna(titanic["age"].mean())
titanic["embarked"] = titanic["embarked"].fillna(titanic["embarked"].mode()[0])
titanic=titanic.drop("deck",axis="columns")
titanic.isnull().sum()
```

## Task4

```python
iris=sns.load_dataset("iris")
print("dimentionality before drop" ,iris.shape)
print("Duplicated entries",iris.duplicated().sum())
iris=iris.drop_duplicates()
print("dimentionality after drop" ,iris.shape)
print("As the 1 duplicated entry is removed dimension reduces to 149,5 from 150,5")

tips=sns.load_dataset("tips")
print("dimentionality before drop" ,tips.shape)
print("Duplicated entries",tips.duplicated().sum())
tips=tips.drop_duplicates()
print("dimentionality after drop" ,tips.shape)

titanic=sns.load_dataset("titanic")
print("dimentionality before drop" ,titanic.shape)
print("Duplicated entries",titanic.duplicated().sum())
titanic=titanic.drop_duplicates()
print("dimentionality after drop" ,titanic.shape)
```

## Task 7

```python
tips = sns.load_dataset("tips")
Q1= tips[["total_bill", "tip"]].quantile(0.25)
Q2= tips[["total_bill", "tip"]].quantile(0.5)
Q3= tips[["total_bill", "tip"]].quantile(0.75)
IQR= Q3 - Q1
low= Q1 - 1.5 * IQR
up= Q3 + 1.5 * IQR
outliers = ((tips[["total_bill", "tip"]]<low)|(tips[["total_bill", "tip"]]>up))
outliers.sum()
sns.boxplot(data=tips[["total_bill","tip"]])
```
